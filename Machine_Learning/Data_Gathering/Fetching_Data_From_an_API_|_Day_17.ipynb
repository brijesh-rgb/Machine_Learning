```python
#Automatically create a DataFrame for every feature

#This is a very useful technique:

import pandas as pd

dfs = {}

for key, value in data['scorecard'][0].items():

    if isinstance(value, list):
        dfs[key] = pd.DataFrame(value)

    elif isinstance(value, dict):
        dfs[key] = pd.DataFrame([value])

#Now all DataFrames are stored in the dictionary dfs.

#For example:

print(dfs['batsman'])
print(dfs['bowler'])
print(dfs['extras'])
print(dfs['fow'])
```
