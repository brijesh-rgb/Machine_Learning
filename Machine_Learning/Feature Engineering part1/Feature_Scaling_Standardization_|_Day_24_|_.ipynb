```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Social_Network_Ads.csv')
```

```python
df=df.iloc[:,2:]
```

```python
df.sample(5)
```

# Train test split

```python
from sklearn.model_selection import train_test_split
x=df[['Age','EstimatedSalary']]
y=df['Purchased']
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
X_test.shape , X_train.shape
```

# StandardScaler

### Step 1 — Import StandardScaler

```python
from sklearn.preprocessing import StandardScaler
```

### Step 2 — Create the scaler object

```python
scaler = StandardScaler()
```

### Step 3 — Fit the scaler on X_train
This is very important.

The scaler calculates the mean and standard deviation of each feature using only X_train.

#### For your data, it learns:

* Age              → mean + standard deviation
* EstimatedSalary  → mean + standard deviation

```python
# fit the scaler to the train set, it will learn the parameters
scaler.fit(X_train)
```

### Step 4 — Transform X_train

```python

# transform train and test sets
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```
